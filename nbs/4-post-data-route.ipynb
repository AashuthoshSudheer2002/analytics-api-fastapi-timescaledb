```python
import requests
```

```python
import json 
data = {"description": "test event",
        "id" : 1,
        "page": "text../"}
```

```python
path = '/api/events/'
endpoint = f"http://localhost:8802{path}"
headers = {
    'Content-Type': 'application/json'
}
response = requests.post(endpoint,data=json.dumps(data),headers=headers)
print(response.headers)
print('ok',response.ok)
if response.ok:
    data = response.json()
    print(data, type(data), data.get('items'))
else:
    print(response.text)

```

```python
detail_path = '/api/events/1'
base_url = 'http://localhost:8802'
detail_endpoint = f"{base_url}{detail_path}"
headers = {
    'Content-Type': 'application/json'
}
response = requests.put(detail_endpoint,json={"description":"Newly updated description!"},headers=headers)
print(response.headers)
print('ok',response.ok)
if response.ok:
    data = response.json()
    print(data, type(data), data.get('id'))
else:
    print(f"text: {response.text}, status code: {response.status_code}")
```
